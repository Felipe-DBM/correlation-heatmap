```python
# Library imports
import pandas as pd # For data manipulation
import seaborn as sns # For data visualization

# Read data from an Excel file
df = pd.read_excel(
    'data.xlsx', # Path to the Excel file 
    sheet_name=4, # Specify the sheet index (0-based) 
    index_col=0 # Assuming the first column contains row labels
    )

# Create a heatmap visualization
sns.heatmap(
    df, # DataFrame to visualize
    center=0.5, # Center the colormap at 0.5
    annot=True, # Annotate each cell with its value
    linewidths=0.9, # Width of the lines that will divide each cell
    linecolor='black', # Color of the lines dividing cells
    cmap='YlGnBu' # Colormap to use for the heatmap
) 
```
